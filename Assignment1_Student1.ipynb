## Section 2 : Define the ML Task

Before writing any model code, you need to be precise about:
1. What are you trying to **predict**? → the **target** variable `y`
2. What information are you allowed to use? → the **features** `X`


**Task 2.1** : Assign the target variable.

The column `mpg` is what we want to predict.  
Fill in the blank so that `y` holds the target series.


```python
# Define the target variable
target = "mpg"           # <-- fill in the column name

y = df[target]
print("Target:", target)
print("y — first 5 values:", y.values[:5])
```

**Task 2.2** : Build the feature matrix `X`.

Fill in the blank to drop the target variable.


```python
# Define the feature matrix
# Drop the target column so it cannot 'leak' into the model
X = df.drop(columns=["mpg"])   # <-- fill in the column name to drop

print("Feature columns:", list(X.columns))
print("X shape:", X.shape)
```

## Section 3 : Train / Test Split

We split the data **once** before touching any model.  
- 80 % for training, 20 % for testing  
- Fixed seed so everyone gets identical splits

**Do not change `test_size` or `random_state`.**


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=SEED
)

print(f"Training samples : {len(X_train)}")
print(f"Test samples     : {len(X_test)}")
```

**Task 3.1** : What is the mean `mpg` in the **training set**?  
Print it rounded to 2 decimal places.


```python
# Compute and print mean mpg in the training set
mean_mpg_train = y_train.mean()
print(f"Mean mpg (train): {mean_mpg_train:.2f}")
```

## Section 4 : Baseline Model

Before trying a "real" model, we always establish a **baseline** — the simplest possible predictor.

A **mean baseline** always predicts the mean of the training target,  
regardless of any feature values.

**Task 4.1** : Fill in the strategy so the baseline always predicts the training mean.


```python
# Create and fit the baseline model
baseline = DummyRegressor(strategy="mean")   # <-- fill in the strategy

baseline.fit(X_train, y_train)

# Predict on the TEST set
y_pred_baseline = baseline.predict(X_test)

print("Baseline always predicts:", round(baseline.constant_[0][0], 4))
```

**Task 4.2** : Compute MAE and RMSE for the baseline on the test set.  
Fill in the blanks.


```python
# Evaluate baseline
baseline_mae  = mean_absolute_error(y_test, y_pred_baseline)
baseline_rmse = np.sqrt(mean_squared_error(y_test, y_pred_baseline))

print(f"Baseline MAE  : {baseline_mae:.4f}")
print(f"Baseline RMSE : {baseline_rmse:.4f}")
```

## Section 5 : Linear Regression Model

Now fit a simple **linear regression** model and compare its error to the baseline.

**Task 5.1** : Train the model on the training data.


```python
# Create and fit a LinearRegression model
lr = LinearRegression()
lr.fit(X_train, y_train)   # <-- fill in (training features, training target)
```

**Task 5.2** : Generate predictions on the **test set** and evaluate.


```python
# Predict on the test set
y_pred_lr = lr.predict(X_test)

# Compute metrics
lr_mae  = mean_absolute_error(y_test, y_pred_lr)
lr_rmse = np.sqrt(mean_squared_error(y_test, y_pred_lr))

print(f"Linear Regression MAE  : {lr_mae:.4f}")
print(f"Linear Regression RMSE : {lr_rmse:.4f}")
```
